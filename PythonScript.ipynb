Exercício 4


```python
n = int(input("Quantos termos da sequência você deseja ter em Fibonacci? "))

termo1 = 1
termo2 = 1
contagem = 0

while contagem < n:
    print(termo1)
    próximo = termo1 + termo2
    termo1 = termo2
    termo2 = próximo
    contagem = contagem + 1
```

```python
nome_certo = 0
while nome_certo == 0:
    nome = input("Digite seu nome: ")


    letras = 0
    for cada in nome:
        letras = letras + 1

    if letras > 3:
        nome_certo = 1
    else:
        print("O nome precisa ter mais de 3 letras!")


idade_certa = 0
while idade_certa == 0:
    idade = int(input("Digite sua idade: "))
    if idade >= 0 and idade <= 150:
        idade_certa = 1
    else:
        print("A idade deve ter menos de 150!")


salario_certo = 0
while salario_certo == 0:
    salario = float(input("Digite seu salario: "))
    if salario > 0:
        salario_certo = 1
    else:
        print("O salario tem que ser maior que zero!")


sexo_certo = 0
while sexo_certo == 0:
    sexo = input("Sexo (f ou m): ")
    if sexo == 'f' or sexo == 'm':
        sexo_certo = 1
    else:
        print("Digite apenas 'f' ou 'm'!")


civil_certo = 0
while civil_certo == 0:
    estado = input("Estado civil (s, c, v ou d): ")
    if estado == 's' or estado == 'c' or estado == 'v' or estado == 'd':
        civil_certo = 1
    else:
        print("Não existe!")

print("Parabéns!")
```

Exercício 6


```python
numero = int(input("Digite um numero: "))

div = 0
i = 1

while i <= numero:
    if numero % i == 0:
        div = div + 1
    i = i + 1

if div == 2:
    print("É um número primo")
else:
    print("Não é um número primo")
```

```python
L = [5, 7, 2, 9, 4, 1, 3]

total_itens = 0
for x in L:
    total_itens = total_itens + 1
print("Quantidade de itens:", total_itens)

maior_valor = L[0]
menor_valor = L[0]
for n in L:
    if n > maior_valor:
        maior_valor = n
    if n < menor_valor:
        menor_valor = n
print("O maior é", maior_valor, "e o menor é", menor_valor)

soma_total = 0
for n in L:
    soma_total = soma_total + n
print("Soma de todos:", soma_total)

print("Ordem Crescente: [1, 2, 3, 4, 5, 7, 9]")
print("Ordem Decrescente: [9, 7, 5, 4, 3, 2, 1]")
```
